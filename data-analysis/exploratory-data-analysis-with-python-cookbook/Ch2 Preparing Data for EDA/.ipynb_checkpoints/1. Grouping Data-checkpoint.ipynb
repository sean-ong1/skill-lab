#### Import relevant libraries

```python
import numpy as np
import pandas as pd
```

#### Load dataset

```python
marketing_data = pd.read_csv("data/marketing_campaign.csv")
```

#### Subset for relevant columns

```python
marketing_data = marketing_data[['ID', 'Year_Birth', 'Education', 'Marital_Status',
                'Income', 'Kidhome', 'Teenhome', 'Dt_Customer', 
                'Recency','NumStorePurchases', 'NumWebVisitsMonth']]
```

#### Check the average number of store purchases of customers based on number of kids in the home

```python
marketing_data.groupby('Kidhome')['NumStorePurchases'].mean()
```

```python
marketing_data.groupby(['Kidhome','Teenhome']).agg(
    NumStorePurchases_sum=('NumStorePurchases', 'sum'),
    NumWebVisitsMonth_median=('NumWebVisitsMonth', 'median')
)
```

```python
marketing_data.groupby(['Kidhome','Teenhome'])[['NumStorePurchases', 'NumWebVisitsMonth']].agg(
    [np.sum, np.mean, np.std]
)
```

```python
marketing_data.groupby(['Kidhome','Teenhome']).agg(
    {'NumStorePurchases': [np.sum, np.size], 'NumWebVisitsMonth': np.mean}
)
```

Transformation methods return a new DataFrame with the same shape and indexing as the original one but with transformed individual values. Here it is important to note that the transformation must not modify any values in the original DataFrame, meaning that such operations cannot be performed in place.

The most common pandas method to transform the data of a GroupBy object is transform(). For example, it can be helpful for computing the z-score for each group:

```python
marketing_data.groupby(['Kidhome','Teenhome'])[['NumStorePurchases', 'NumWebVisitsMonth']].transform(lambda x: (x - x.mean()) / x.std())
```

```python
marketing_data.groupby(['Kidhome','Teenhome'])[['NumStorePurchases', 'NumWebVisitsMonth']].filter(
    lambda x: x['NumStorePurchases'].mean() > 5 and x['NumWebVisitsMonth'].mean() > 5)
```

```python
marketing_data.groupby(['Kidhome','Teenhome']).nth(3)
```
