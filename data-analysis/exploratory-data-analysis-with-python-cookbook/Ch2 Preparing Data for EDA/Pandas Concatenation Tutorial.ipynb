```python
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
north_america = pd.read_csv('./Data/PandasConcatTutorial/Pandas Concatenation Tutorial/north_america_2000_2010.csv', index_col=0)
south_america = pd.read_csv('./Data/PandasConcatTutorial/Pandas Concatenation Tutorial/south_america_2000_2010.csv', index_col=0)
```

```python
north_america
```

By default, the DataFrame.plot() method treats rows as x axis labels, cell values as y axis labels, and columns as lines.

```python
north_america.transpose().plot(title='Average Labor Hours Per Year')
plt.show()
```

```python
south_america.transpose().plot(title='Average Labor Hours Per Year')
plt.show()
```

```python
americas = pd.concat([north_america, south_america])
americas
```

```python
americas_dfs = [americas]
for year in range(2011, 2016):
    filename = "./Data/PandasConcatTutorial/Pandas Concatenation Tutorial/americas_{}.csv".format(year)
    df = pd.read_csv(filename, index_col=0)
    americas_dfs.append(df)
americas_dfs[1]
```

```python
americas = pd.concat(americas_dfs, axis=1)
# americas.index.names = ['Country']
americas
```

```python
americas.transpose().plot(title='Average Labor Hours Per Year')
plt.show()
```

```python
asia = pd.read_csv('./Data/PandasConcatTutorial/Pandas Concatenation Tutorial/asia_2000_2015.csv', index_col=0)
asia
```

```python
europe = pd.read_csv('./Data/PandasConcatTutorial/Pandas Concatenation Tutorial/europe_2000_2015.csv', index_col=0)
europe.head()
```

```python
south_pacific = pd.read_csv('./Data/PandasConcatTutorial/Pandas Concatenation Tutorial/south_pacific_2000_2015.csv', index_col=0)
south_pacific
```

```python
world = pd.concat([americas, asia, europe, south_pacific])
world.index
```

```python
world.transpose().plot(figsize=(10,10), colormap='rainbow', linewidth=2, title='Average Labor Hours Per Year')
plt.legend(loc='right', bbox_to_anchor=(1.3, 0.5))
plt.show()
```

```python
historical = pd.read_csv('./Data/PandasConcatTutorial/Pandas Concatenation Tutorial/historical.csv', index_col=0)
historical.head()
```

```python
world_historical = pd.merge(historical, world, left_index=True, right_index=True, how='right') # indexes will be key value
print(world_historical.shape)
world_historical.head()
```

```python
world_historical = historical.join(world, how='right') # default on=None (caller index), on refers caller's column, right dataframe alyways on index
world_historical.head()
```

```python
world_historical.sort_index(inplace=True)
world_historical.transpose().plot(figsize=(15,10), colormap='rainbow', linewidth=2, title='Average Labor Hours Per Year')
plt.legend(loc='right', bbox_to_anchor=(1.15, 0.5))
plt.show()
```
