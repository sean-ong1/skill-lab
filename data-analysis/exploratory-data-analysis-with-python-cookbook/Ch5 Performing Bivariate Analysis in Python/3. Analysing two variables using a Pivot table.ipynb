```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
penguins_data = pd.read_csv("data/penguins_size.csv")
pd.pivot_table(penguins_data, values=['culmen_length_mm', 'body_mass_g'], 
                                index='species',  
                                aggfunc=['mean', 'min', 'max'])
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
penguins_data = pd.read_csv("data/penguins_size.csv")
pd.pivot_table(penguins_data, values=['culmen_length_mm', 'body_mass_g'], 
                                index='species',  
                                columns='sex',
                                aggfunc={'culmen_length_mm': "mean",'body_mass_g': ['mean', 'min', 'max']})
```
