```python
import pandas as pd
from statsmodels.tsa.stattools import kpss

air_traffic_data = pd.read_csv("data/SF_Air_Traffic_Passenger_Statistics_Transformed.csv")
air_traffic_data['Date']= pd.to_datetime(air_traffic_data['Date'], format = "%Y%m")
air_traffic_data.set_index('Date',inplace = True)
# regression='c' (Constant): Assumes the data fluctuates around a horizontal baseline.
# regression='ct' (Constant and Trend): Assumes the data fluctuates around a sloped baseline.
result_kpss_ct=kpss(air_traffic_data,regression="c")
print(f'Test Statistic: {result_kpss_ct[0]}')
print(f'p-value: {result_kpss_ct[1]}')
print(f'Critical values:')
for key, value in result_kpss_ct[3].items():
     print(f'\t{key}: {value:.3f}')

if result_kpss_ct[0] > result_kpss_ct[3]["5%"] or result_kpss_ct[1] < 0.05:
    print (f'Reject Null Hypothesis - Time Series is Non-Stationary')
else:
    print (f'Failed to Reject Null Hypothesis - Time Series is Stationary')
```
